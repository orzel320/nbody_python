```python
from nbody_sim.api import Simulation
from nbody_sim.render import create_k3d_animation

sim, config = Simulation.from_json('scenarios/asteroid_belt.json')

name = config.get('name', 'Simulation')
print(f"Baking high-performance K3D projection for {name}. Please wait...")

plot = create_k3d_animation(
    sim, 
    dt=config['dt'], 
    sizes=config['sizes'], 
    colours=config['colours'], 
    steps=300 
)

plot.display()

plot.camera = [0, -15, 10,  0, 0, 0,  0, 0, 1]
```

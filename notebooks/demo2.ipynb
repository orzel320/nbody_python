```python
import numpy as np
from nbody_sim.api import Simulation, ureg
from nbody_sim.render import create_k3d_animation

print("Initializing N-Body Physics Engine...")
sim = Simulation()

# 1. Generate 500 procedural asteroids
np.random.seed(42)
n_bodies = 500

# Using your Pint unit registry for strict physical definitions
positions = np.random.uniform(-5.0, 5.0, (n_bodies, 3)) * ureg.astronomical_unit
velocities = np.random.uniform(-0.5, 0.5, (n_bodies, 3)) * (ureg.astronomical_unit / ureg.nbody_time)
masses = np.random.uniform(0.1, 2.0, n_bodies) * ureg.solar_mass

# Add them to the Barnes-Hut tree
sim.add_bodies(positions, velocities, masses)

# Generate procedural colours (RGBA floats) and sizes for the renderer
colours = np.random.uniform(0.4, 1.0, (n_bodies, 4))
colours[:, 3] = 1.0 # Set full opacity
sizes = np.random.uniform(0.5, 2.0, n_bodies)

print("Baking 3D WebGL animation. Please wait...")

# 2. Render the interactive K3D plot
plot = create_k3d_animation(
    sim,
    dt=0.01,
    sizes=sizes,
    colours=colours,
    steps=150
)

# Display the widget!
plot.display()
```
